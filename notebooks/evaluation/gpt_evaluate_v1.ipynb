# LLM-as-a-judge — GPT-4o / FOA paper protocol v2

Bật Internet; bật GPU cho attack/caption local. Thêm repository và dữ liệu vào Kaggle Inputs.
Dùng một CONFIG_NAME chung cho mọi model captioning: evaluate_gpt4o_judge_v2. Chỉ đổi INPUT_DIR tới một run caption.
Judge cố định gpt-4o; model captioning được đọc từ captions.jsonl, không được dùng làm judge. ASR dùng score > 0.5; AvgSim là điểm target–adversarial trung bình.
Chỉ cần Secret OPENAI_API_KEY. Evaluation chỉ gửi hai caption dạng text, không chạy lại model captioning và không cần GPU.
Mọi xử lý nằm trong các file Python hiện có. Xem src/FOA-Attack/config/README.md.



```python
from pathlib import Path
import subprocess
import sys

REPO_DIR = Path("/kaggle/input/foa-code/Code-Adversarial-Attack-on-VLM")
CONFIG_NAME = "evaluate_gpt4o_judge_v2"
OUTPUT_DIR = "/kaggle/working/foa_outputs"
INPUT_DIR = "/kaggle/input/foa-caption-output"
RUNNER = REPO_DIR / "src" / "FOA-Attack" / "utils.py"
assert RUNNER.is_file(), f"Chỉnh REPO_DIR: không tìm thấy {RUNNER}"
```

```python
subprocess.run([sys.executable, "-m", "pip", "install",
                "hydra-core==1.3.2", "omegaconf==2.3.0", "PyYAML==6.0.2", "wandb==0.19.1"], check=True)
subprocess.run([sys.executable, str(RUNNER), "setup", "--config-name", CONFIG_NAME], check=True)
```

```python
command = [sys.executable, str(RUNNER), "run", "--config-name", CONFIG_NAME, "--output", OUTPUT_DIR]
command += ["--input", INPUT_DIR]
subprocess.run(command, check=True)
```

```python
subprocess.run([sys.executable, str(RUNNER), "export", "--config-name", CONFIG_NAME, "--output", OUTPUT_DIR], check=True)
from IPython.display import FileLink, display
for archive in sorted(Path(OUTPUT_DIR).glob(f"{CONFIG_NAME}_*.zip")):
    display(FileLink(str(archive)))
```
